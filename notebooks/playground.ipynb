```python
import torch
import matplotlib.pyplot as plt
from torchvision import datasets, transforms

# Load corrupted
corrupt_images = torch.load('../data/corruptmnist_v1/train_images_0.pt')

# Load original MNIST
transform = transforms.Compose([transforms.ToTensor()])
mnist = datasets.MNIST(root='../data', train=True, download=True, transform=transform)

# Plot side by side
for i in range(9):
    plt.subplot(9, 2, 2*i + 1)
    plt.imshow(mnist[i][0].squeeze(), cmap='gray')
    plt.axis('off')
    if i == 0:
        plt.title('Original')
    
    plt.subplot(9, 2, 2*i + 2)
    plt.imshow(corrupt_images[i], cmap='gray')
    plt.axis('off')
    if i == 0:
        plt.title('Corrupted')

plt.show()
```
